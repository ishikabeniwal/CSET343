```python
import pandas as pd

data = {
    "Name": ["Ishika", "Rahul", "Aman"],
    "Age": [21, 22, 20],
    "Marks": [91, 85, 88]
}

df = pd.DataFrame(data)

print(df)
```

```python
# First 2 rows
print(df.head(2))

# Column selection
print(df["Name"])

# Average marks
print(df["Marks"].mean())

# Students with marks above 87
print(df[df["Marks"] > 87])
```

```python
import numpy as np

arr = np.array([10, 20, 30, 40])

print(arr)
print("Mean:", np.mean(arr))
print("Sum:", np.sum(arr))
print("Maximum:", np.max(arr))
```

```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

print("Addition:\n", A + B)
print("Multiplication:\n", A @ B)
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
sales = [100, 150, 130, 170, 200]

plt.plot(days, sales, marker="o")
plt.title("Sales Over Days")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["A", "B", "C", "D"]
marks = [80, 90, 75, 95]

plt.bar(students, marks)
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```
